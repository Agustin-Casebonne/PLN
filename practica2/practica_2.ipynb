# 2. Construcción bag of words

Primero juntamos todas las palabras en un mismo vector y las ordenamos. Segundo para la palabra lo que hacemos es recorrer cada documento y se extrae del vector anterior cada palabra para contarla en cada documento.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Vocabulario: todas las palabras distintas de todos los documentos, ordenadas
vocabulario = set()
for palabras in dicc.values():
    vocabulario.update(palabras)
vocabulario = sorted(vocabulario)
print("Tamaño del vocabulario:", len(vocabulario))

# 2. Bag-of-Words: una fila por documento, una columna por palabra del vocabulario.
#    Cada valor es cuántas veces aparece esa palabra en ese documento.
bow = []
for nombre in nombres: #keys() del diccionario
    palabras_doc = dicc[nombre] #obtener las palabras del documento
    fila = []
    for palabra in vocabulario:
        fila.append(palabras_doc.count(palabra)) #contamos cuántas veces aparece cada palabra del vocabulario en el documento
    bow.append(fila)

bow = np.array(bow)
df_bow = pd.DataFrame(bow, index=nombres, columns=vocabulario)
df_bow.iloc[:, :15]
```

# 3. Similitud del Coseno

Medir la similitud entre documentos 1 iguales, 0 sin similitud(ortogonales) y -1 opuestos.

```python
import torch
from torch.nn.functional import cosine_similarity

"""def similitud_coseno(a, b):
    # Fórmula: (a · b) / (|a| * |b|)
    producto_punto = np.dot(a, b)        # suma de multiplicar posición a posición
    norma_a = np.sqrt(np.sum(a * a))     # "longitud" del vector a
    norma_b = np.sqrt(np.sum(b * b))     # "longitud" del vector b
    return producto_punto / (norma_a * norma_b)
"""



n = len(nombres)
sim = np.zeros((n, n))                   # tabla vacía de 12x12

for i in range(n):
    for j in range(n):
        sim[i, j] = cosine_similarity(
            torch.tensor(bow[i], dtype=torch.float32).unsqueeze(0), 
            torch.tensor(bow[j], dtype=torch.float32).unsqueeze(0)
        ).item()

df_sim = pd.DataFrame(sim, index=nombres, columns=nombres)
df_sim.round(3)
```

```python
plt.figure(figsize=(8, 6))
plt.imshow(sim, cmap="viridis", vmin=0, vmax=1)   # cada casilla es un color según su valor
plt.colorbar(label="Similitud del coseno")

# Nombres de los documentos en los ejes
plt.xticks(range(n), nombres, rotation=90)
plt.yticks(range(n), nombres)

# Escribir el número dentro de cada casilla
for i in range(n):
    for j in range(n):
        plt.text(j, i, round(sim[i, j], 2), ha="center", va="center", color="white", fontsize=7)

plt.title("Similitud del coseno entre documentos")
plt.tight_layout()
plt.savefig("heatmap_similitud.png", dpi=150)
plt.show()
```

# 4. Análisis

Calcular nuevamente el valor de similitud entre documentos

```python
# Armar la lista de todos los pares de documentos, sin repetir (doc1-doc2 es igual que doc2-doc1)
pares = []
for i in range(n):
    for j in range(i + 1, n):
        pares.append([nombres[i], nombres[j], sim[i, j]]) # agregar el par de documentos y su similitud

df_pares = pd.DataFrame(pares, columns=["doc_a", "doc_b", "similitud"])
df_pares = df_pares.sort_values("similitud", ascending=False)

print("Pares más similares")
display(df_pares.head(10).round(3))
print("Pares menos similares")
display(df_pares.tail(10).round(3))
```

Para ver si hablan de temas similares vamos identificar las palabras del texto y ver cuales aparecen

```python
palabras_movil = ["móvil", "móviles", "smartphone", "smartphones", "aplicación", "aplicaciones",
                   "app", "apps", "teléfono", "teléfonos", "celular", "pantalla", "cámara", "procesador"]
palabras_ev = ["eléctrico", "eléctricos", "eléctrica", "eléctricas", "vehículo", "vehículos",
               "batería", "baterías", "autonomía", "carga", "coche", "coches", "automóvil", "automóviles"]

temas = {}
for nombre in nombres:
    palabras_doc = dicc[nombre]

    tiene_movil = False
    for w in palabras_doc:
        if w in palabras_movil:
            tiene_movil = True

    tiene_ev = False
    for w in palabras_doc:
        if w in palabras_ev:
            tiene_ev = True

    if tiene_movil and tiene_ev:
        temas[nombre] = "ambos"
    elif tiene_movil:
        temas[nombre] = "móvil"
    elif tiene_ev:
        temas[nombre] = "vehículos eléctricos"
    else:
        temas[nombre] = "otro"

print(pd.Series(temas, name="tema"))
```

La siguiente celda sirve para añadir a cada pareja de documentos la información de si pertenecen al mismo tema:

```python
lista_mismo_tema = []

for fila in df_pares.itertuples():
    tema_a = temas[fila.doc_a]
    tema_b = temas[fila.doc_b]

    lista_mismo_tema.append(tema_a == tema_b)

df_pares["mismo_tema"] = lista_mismo_tema
```

```python
similitudes_mismo_tema = df_pares[
    df_pares["mismo_tema"] == True
]["similitud"]

similitudes_distinto_tema = df_pares[
    df_pares["mismo_tema"] == False
]["similitud"]
```

```python
print("Mismo tema:")
print("Media:", round(similitudes_mismo_tema.mean(), 3))
print("Mínimo:", round(similitudes_mismo_tema.min(), 3))
print("Máximo:", round(similitudes_mismo_tema.max(), 3))
print("Número de parejas:", len(similitudes_mismo_tema))

print("\nTemas distintos:")
print("Media:", round(similitudes_distinto_tema.mean(), 3))
print("Mínimo:", round(similitudes_distinto_tema.min(), 3))
print("Máximo:", round(similitudes_distinto_tema.max(), 3))
print("Número de parejas:", len(similitudes_distinto_tema))
```

### Análisis de similitud según el tema

Para comprobar si los documentos que tratan sobre un mismo tema presentan una mayor similitud, se han separado las 66 parejas de documentos en dos grupos: parejas que pertenecen al mismo tema y parejas que pertenecen a temas diferentes.

De las 66 parejas analizadas, **19 pertenecen al mismo tema**, mientras que **47 corresponden a temas diferentes**. En las parejas del mismo tema, la similitud media es de **0,120**, con un mínimo de **0,000** y un máximo de **0,302**. Por otro lado, las parejas de temas diferentes presentan una similitud media de **0,070**, con un mínimo de **0,000** y un máximo de **0,323**.

Los resultados muestran que la **similitud media es mayor entre los documentos del mismo tema (0,120) que entre documentos de temas diferentes (0,070)**. Esto indica que, en promedio, los documentos que tratan sobre contenidos relacionados comparten una mayor cantidad de vocabulario y, por tanto, sus vectores Bag-of-Words presentan una orientación más parecida.

Sin embargo, esta relación no se cumple para todas las parejas. Por ejemplo, existen documentos del mismo tema con una similitud de **0,000**, mientras que la pareja con mayor similitud de todo el conjunto (**0,323**) pertenece a documentos clasificados como de temas diferentes. Por tanto, la clasificación temática no determina directamente la similitud entre documentos, sino que permite observar una tendencia general: los documentos del mismo tema tienden a ser más similares en promedio.
